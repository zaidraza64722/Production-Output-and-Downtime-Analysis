#### Check timestamp consistency

```python
# 1. Convert interval_start temporarily for inspection
interval_dates = pd.to_datetime(
    df["interval_start"],
    errors="coerce",
    utc=True
)

# 2. Check date ranges
print("interval_start date range:")
print("Minimum:", interval_dates.min())
print("Maximum:", interval_dates.max())

# 3. Check start timestamp range
start_dates = pd.to_datetime(df["start"], unit="s", utc=True)

print("\nStart date range:")
print("Minimum:", start_dates.min())
print("Maximum:", start_dates.max())

# 4. Check end timestamp range
end_dates = pd.to_datetime(df["end"], unit="s", utc=True)

print("\nEnd date range:")
print("Minimum:", end_dates.min())
print("Maximum:", end_dates.max())

# 5. Check invalid interval_start values
print("\nInvalid interval_start values:")
print(interval_dates.isna().sum())
```

##### This means 1,016 values could not be converted into valid timestamps using our parsing operation.

```python
#  Investigate invalid timestamps

# Display rows with invalid interval_start values
invalid_dates = df[interval_dates.isna()]

print("Number of invalid records:", len(invalid_dates))

# Display first 20 invalid values
print("\nInvalid timestamp examples:\n")
print(invalid_dates["interval_start"].head(20))

# Check their machine IDs
print("\nInvalid records by machine:\n")
print(invalid_dates["equipment_ID"].value_counts())

# Check their production types
print("\nInvalid records by production type:\n")
print(invalid_dates["type"].value_counts())
```

##### These look like valid date-time strings. The problem may be that Pandas is interpreting timestamps with different formats inconsistently.
- With fractional seconds: 2020-01-01 11:21:28.907000+00:00
- Without fractional seconds: 2020-01-01 23:00:00+00:00

Pandas may infer one format from the first value and fail to parse values using the other format. 

```python
## Step 1: Try mixed-format parsing

# Parse timestamps while allowing mixed formats
interval_dates_mixed = pd.to_datetime(
    df["interval_start"],
    format="mixed",
    errors="coerce",
    utc=True
)

print("Original parsing failures:", interval_dates.isna().sum())

print("Failures after mixed parsing:", interval_dates_mixed.isna().sum())

print("\nSuccessfully parsed timestamps:")
print(interval_dates_mixed.notna().sum())
```

```python
# Inspect any remaining failures

remaining_invalid = df[interval_dates_mixed.isna()]
print("Remaining invalid records:", len(remaining_invalid))
print(remaining_invalid["interval_start"].head(20))
```

#### Check whether machine counters reset

```python
# Sort records chronologically for each machine
df_sorted = df.sort_values(
    ["equipment_ID", "interval_start"]
).copy()

# Calculate consecutive counter differences
df_sorted["pi_diff"] = (
    df_sorted.groupby("equipment_ID")["pi"].diff()
)
# This calculates the difference between the current PI value and 
# the previous PI value for the same machine.
# Example: 1,500 − 1,000 = 500.

df_sorted["po_diff"] = (
    df_sorted.groupby("equipment_ID")["po"].diff()
)

# Count negative differences
print("Negative PI counter differences:",
      (df_sorted["pi_diff"] < 0).sum())

print("Negative PO counter differences:",
      (df_sorted["po_diff"] < 0).sum())

# Display examples
print("\nPI counter decrease examples:")
print(df_sorted[df_sorted["pi_diff"] < 0][
    ["interval_start", "equipment_ID", "pi", "pi_diff"]
].head(10))

print("\nPO counter decrease examples:")
print(df_sorted[df_sorted["po_diff"] < 0][
    ["interval_start", "equipment_ID", "po", "po_diff"]
].head(10))
```

#### Find counter decreases machine-wise

```python
# Count PI counter decreases for each machine
print("PI Counter Decreases by Machine:")
print(
    df_sorted[df_sorted["pi_diff"] < 0]
    .groupby("equipment_ID")
    .size()
)

# Count PO counter decreases for each machine
print("\nPO Counter Decreases by Machine:")
print(
    df_sorted[df_sorted["po_diff"] < 0]
    .groupby("equipment_ID")
    .size()
)
```

```python
# Show PI counter decreases for machine s_4
s4_pi_resets = df_sorted[
    (df_sorted["equipment_ID"] == "s_4") &
    (df_sorted["pi_diff"] < 0)
]

print("Machine s_4 - PI counter decreases:")
print(
    s4_pi_resets[
        ["interval_start", "pi", "pi_diff"]
    ].head(15)
)

# Show PO counter decreases for machine s_4
s4_po_resets = df_sorted[
    (df_sorted["equipment_ID"] == "s_4") &
    (df_sorted["po_diff"] < 0)
]

print("\nMachine s_4 - PO counter decreases:")
print(
    s4_po_resets[
        ["interval_start", "po", "po_diff"]
    ].head(15)
)
```

#### Check the operating types during counter decreases

```python
# Get machine s_4 records where PI counter decreased
s4_pi_resets = df_sorted[
    (df_sorted["equipment_ID"] == "s_4") &
    (df_sorted["pi_diff"] < 0)
]

# Check production types during these records
print("Production types during PI decreases:")
print(s4_pi_resets["type"].value_counts())

# Check alarm codes during these records
print("\nAlarm codes during PI decreases:")
print(s4_pi_resets["alarm"].value_counts().head(10))
```

## Data Cleaning & Preparation

```python
# Create a working copy
df_clean = df.copy()

# Rename columns for better understanding
df_clean.rename(columns={
    "interval_start": "record_datetime",
    "equipment_ID": "machine_id",
    "alarm": "alarm_code",
    "type": "operating_status",
    "start": "event_start_unix",
    "end": "event_end_unix",
    "elapsed": "duration_ms",
    "pi": "input_packages",
    "po": "output_packages",
    "speed": "production_speed"
}, inplace=True)

print(df_clean.columns.tolist())
```

#### Convert record date into datetime format

```python
df_clean["record_datetime"] = pd.to_datetime(
    df_clean["record_datetime"],
    format="mixed",
    errors="coerce",
    utc=True
)

print(df_clean["record_datetime"].dtype)
```

```python
# Convert event timestamps into readable dates

df_clean["event_start"] = pd.to_datetime(
    df_clean["event_start_unix"], unit="s", utc=True
)

df_clean["event_end"] = pd.to_datetime(
    df_clean["event_end_unix"], unit="s", utc=True
)
```

```python
df_clean[["event_start", "event_end"]].head()
```

#### Create duration in minutes

```python
df_clean["duration_minutes"] = df_clean["duration_ms"] / 60000
```

```python
df_clean[["duration_ms", "duration_minutes"]].head()
```

#### Validate the cleaned dataset

```python
print("Missing values after preprocessing:")
print(df_clean.isnull().sum())
```

#### Check duplicate records

```python
print("Duplicate records:", df_clean.duplicated().sum())
```
